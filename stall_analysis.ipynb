## Introduction

This notebook analyses Wanjiku's daily sales over a period of 14 days and evaluates her current stock levels. Using core Python, the analysis identifies sales trends, compares weekday and weekend performance, classifies daily sales, and determines which products need restocking. The findings will help Wanjiku make informed business decisions.


```python

daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```


## Task 1: Daily Sales Statistics

In this task, I calculate the total and average sales for the 14-day period. I also identify the highest and lowest sales amounts and determine the corresponding day names and week numbers using list indexing.


```python

# Calculate total and average sales
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

# Find the highest and lowest sales
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the positions of the highest and lowest sales
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

# Identify the day names
best_day = days[best_index]
worst_day = days[worst_index]

# Identify the week numbers
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

# Display the results
print("DAILY SALES STATISTICS")
print("----------------------")
print(f"Total Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,}")
print(f"Best Day: {best_day}, Week {best_week}")
print(f"Lowest Sales: KES {lowest_sales:,}")
print(f"Worst Day: {worst_day}, Week {worst_week}")
```


## Task 2: Weekly Sales Totals and Growth

In this task, I create a reusable function called week_totals() to calculate sales totals for each week. I then use the weekly totals to determine the percentage growth or decline in sales from Week 1 to Week 2.


```python

def week_totals(sales):
    """
    Calculate and return the sales totals for Week 1 and Week 2.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2


# Calculate weekly totals
week1_total, week2_total = week_totals(daily_sales)

# Calculate week-on-week growth percentage
growth = ((week2_total - week1_total) / week1_total) * 100

# Display results
print("WEEKLY SALES ANALYSIS")
print("---------------------")
print(f"Week 1 Sales: KES {week1_total:,}")
print(f"Week 2 Sales: KES {week2_total:,}")
print(f"Week-on-Week Growth: {growth:.2f}%")
```

```python

def classify_day(amount, average):
    """
    Classify daily sales as Strong, Normal or Weak
    based on the average daily sales.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


# Create a dictionary to count each category
sales_categories = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Classify each day's sales
for sale in daily_sales:
    category = classify_day(sale, average_sales)
    sales_categories[category] += 1

# Display the classification results
print("DAILY SALES CLASSIFICATION")
print("--------------------------")

for category, count in sales_categories.items():
    print(f"{category}: {count} days")
```

```python

print("CHECKING EACH DAY")
print("------------------")

for i in range(len(daily_sales)):
    category = classify_day(daily_sales[i], average_sales)
    print(f"{days[i]}: KES {daily_sales[i]:,} - {category}")

print("\nAverage Sales:", average_sales)
print("Strong threshold:", average_sales * 1.20)
print("Weak threshold:", average_sales * 0.80)
```

## Task 4: Weekday vs Weekend Performance

In this task, I separate sales into weekdays (Monday to Friday) and weekends (Saturday and Sunday). I calculate the average sales for each group and determine the percentage difference to identify whether the stall performs better on weekdays or weekends.

```python
# Separate weekday and weekend sales
weekday_sales = []
weekend_sales = []

for day, sale in zip(days, daily_sales):
    if day in ["Sat", "Sun"]:
        weekend_sales.append(sale)
    else:
        weekday_sales.append(sale)

# Calculate average sales
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

# Display results
print("WEEKDAY VS WEEKEND PERFORMANCE")
print("--------------------------------")
print(f"Average Weekday Sales: KES {weekday_average:,.2f}")
print(f"Average Weekend Sales: KES {weekend_average:,.2f}")
print(f"Percentage Difference: {percentage_difference:.2f}%")
```

## Task 5: Stock Monitoring and Reorder Alerts

In this task, I examine the current stock levels of each product and compare them with their reorder levels. I use conditional statements to identify products that need restocking and calculate the additional quantity required.

```python
# Check stock levels and identify products needing restocking

print("STOCK MONITORING REPORT")
print("-----------------------")

for product, details in stock.items():
    quantity = details["qty"]
    reorder_level = details["reorder"]

    if quantity < reorder_level:
        shortage = reorder_level - quantity
        print(f"{product}: RESTOCK REQUIRED")
        print(f"Current Stock: {quantity} kg")
        print(f"Reorder Level: {reorder_level} kg")
        print(f"Additional Quantity Needed: {shortage} kg")
        print("-----------------------")
    else:
        print(f"{product}: Stock is sufficient")
        print("-----------------------")
```

## Task 6: Product Profit Analysis

In this task, I calculate the profit per kilogram for each product by subtracting its buying cost from its selling price. I then estimate the total potential profit based on the current stock quantities.

```python
# Calculate profit for each product

print("PRODUCT PROFIT ANALYSIS")
print("-----------------------")

total_potential_profit = 0

for product, details in stock.items():
    quantity = details["qty"]
    cost = details["cost"]
    price = details["price"]

    profit_per_kg = price - cost
    total_profit = profit_per_kg * quantity

    total_potential_profit += total_profit

    print(f"{product}:")
    print(f"Profit per kg: KES {profit_per_kg:,}")
    print(f"Potential profit: KES {total_profit:,}")
    print("-----------------------")

print(f"Total Potential Profit: KES {total_potential_profit:,}")
```

## Task 7: Reorder Priority

In this task, I create a function to identify products that need restocking and arrange them according to their stock shortages. Products with larger shortages are given higher priority.

```python
def reorder_priority(stock):
    """
    Identify products below their reorder levels
    and arrange them by shortage from highest to lowest.
    """
    reorder_list = []

    for product, details in stock.items():
        shortage = details["reorder"] - details["qty"]

        if shortage > 0:
            reorder_list.append((product, shortage))

    reorder_list.sort(key=lambda item: item[1], reverse=True)

    return reorder_list


# Display reorder priorities
print("REORDER PRIORITY REPORT")
print("-----------------------")

priorities = reorder_priority(stock)

for product, shortage in priorities:
    print(f"{product}: Restock {shortage} kg")
```
